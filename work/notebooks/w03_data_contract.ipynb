## 3. Verify it with queries (grain, counts, missing values, windows)

*Every claim above gets a query cell here. A contract claim without a query next to it is a guess.*

The dataset should contain one row per content item for a client and reporting date. I will verify the row count, duplicate keys, missing values, and available date range using the dataset itself.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd

# Load dataset
df = pd.read_csv("content_refresh_anonymized.csv")

print("Dataset shape:", df.shape)

# 1. Check duplicate rows
print("\nDuplicate rows:", df.duplicated().sum())

# 2. Show columns
print("\nColumns:")
print(df.columns.tolist())

# 3. Missing values
print("\nMissing values:")
missing = df.isnull().sum()
print(missing[missing > 0].sort_values(ascending=False))

# 4. Look for date columns
date_columns = [col for col in df.columns if "date" in col.lower()]

print("\nPossible date columns:")
print(date_columns)

# 5. Check date ranges
for col in date_columns:
    try:
        dates = pd.to_datetime(df[col], errors="coerce")
        print(f"\n{col}:")
        print("Minimum:", dates.min())
        print("Maximum:", dates.max())
        print("Valid dates:", dates.notna().sum())
    except:
        pass

# 6. Check possible client/content identifiers
possible_id_columns = [
    col for col in df.columns
    if any(word in col.lower() for word in ["client", "content", "item", "id"])
]

print("\nPossible ID columns:")
print(possible_id_columns)
```
